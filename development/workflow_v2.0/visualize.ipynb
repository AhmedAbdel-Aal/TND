````python
import re

def extract_json_from_text(text):
    # Check if the text already looks like a valid JSON object or array
    text = text.strip()  # Remove leading/trailing whitespaces

    if text.startswith("{") and text.endswith("}"):  # Check for JSON object
        try:
            return json.loads(text)  # Directly load the JSON object
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON: {e}")
            return None
    elif text.startswith("[") and text.endswith("]"):  # Check for JSON array
        try:
            return json.loads(text)  # Directly load the JSON array
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON: {e}")
            return None

    # Regular expression to match the JSON part within the text if not already in valid JSON format
    json_pattern = r'```json\n(.*?)\n```'

    # Search for the JSON part within the text
    match = re.search(json_pattern, text, re.DOTALL)
    
    if match:
        # Extract the JSON string
        json_str = match.group(1)
        
        # Parse the JSON string into a Python dictionary
        try:
            json_object = json.loads(json_str)
            return json_object
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON: {e}")
            return None
    else:
        print("No JSON found in the provided text.")
        return None
````
