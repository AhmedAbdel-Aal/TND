```python
import os
```

```python
def load_txt_file(file_path):
    with open(file_path, 'r') as file:
        return file.read()
```

```python
data = load_txt_file('all.txt')
```

```python
cases = data.split('------------------------------------------------------------------------')
```

```python
def parse_string_to_json(text: str) -> dict:
    # Split text into non-empty lines (remove empty lines, if any).
    lines = [l.strip() for l in text.split('\n') if l.strip()]

    # Prepare placeholders
    article = ""
    sub_article = ""
    header = ""
    body_lines = []

    # At minimum, we expect the first line to be "ARTICLE x"
    if len(lines) >= 1:
        article = lines[0]

    # Start scanning from line index 1
    idx = 1

    # If there's a second line AND it starts with "ARTICLE",
    # treat it as the sub_article
    if idx < len(lines) and lines[idx].upper().startswith("ARTICLE"):
        sub_article = lines[idx]
        idx += 1

    # If there's still a line left, treat it as the header
    if idx < len(lines):
        header = lines[idx]
        idx += 1

    # Everything else is the body
    if idx < len(lines):
        # Join the remaining lines using a newline or space
        # depending on how you prefer to preserve formatting.
        body_lines = lines[idx:]
        line_text = "\n".join(body_lines)
    else:
        line_text = ""

    return {
        "article": article,
        "sub-article": sub_article,
        "header": header,
        "line": line_text
    }


# --- Example usage ---

input_text_1 = """ARTICLE 2
EFFECTIVE INVESTIGATION
Alleged failure to conduct effective investigation into road-traf-
fic accident in which an individual sustained unintentional
life-threatening injuries: Article 2 applicable; no violation
"""

parsed_1 = parse_string_to_json(input_text_1)
print(parsed_1)
# Expected:
# {
#   'article': 'ARTICLE 2',
#   'sub-article': '',
#   'header': 'EFFECTIVE INVESTIGATION',
#   'line': 'Alleged failure to conduct effective investigation into ... no violation'
# }


input_text_2 = """ARTICLE 6
ARTICLE 6 § 1 (criminal)
TRIBUNAL ESTABLISHED BY LAW
Participation of judge whose appointment was vitiated by undue executive
discretion without effective domestic court review and redress: violation
"""

parsed_2 = parse_string_to_json(input_text_2)
print(parsed_2)
# Expected:
# {
#   'article': 'ARTICLE 6',
#   'sub-article': 'ARTICLE 6 § 1 (criminal)',
#   'header': 'TRIBUNAL ESTABLISHED BY LAW',
#   'line': 'Participation of judge whose appointment was vitiated ... violation'
# }    
```

```python
cases_dict = {}
for case in cases:
    splits = case.split('\n\n')
    case_id = splits[0].strip().replace('\'', '')
    reasons = splits[1:]
    reasons = [parse_string_to_json(reason) for reason in reasons]
    
    cases_dict[case_id] = reasons
```

```python
import json
def write_json_file(file_path, data):
    with open(file_path, 'w') as file:
        json.dump(data, file, indent=4)

write_json_file('key_cases_lines.json', cases_dict)
```
