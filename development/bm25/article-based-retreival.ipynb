```python
import os
import json
import pandas as pd
from util import *
```

```python
data_path = '/Users/ahmed/Desktop/msc-24/ECHR/echr-processed'
files = os.listdir(data_path)
```

```python
article_map = {}
for file in files:
    with open(os.path.join(data_path, file), 'r') as f:
        data = json.load(f)
        itemid = data['itemid']
        articles = data['__articles'].split(';')
        conclusion = data['__conclusion'].split(';')
        for article in articles:
            if article not in article_map:
                article_map[article] = [itemid]
            else:
                article_map[article].append(itemid)        
```

```python
def read_data():
    df_1 = pd.read_csv("/Users/ahmed/Desktop/msc-24/TND/kc_classification_data/pre_cutoff_data/df_1.csv")
    list_1 = df_1["file_path"].to_list()

    df_2 = pd.read_csv("/Users/ahmed/Desktop/msc-24/TND/kc_classification_data/pre_cutoff_data/df_2.csv")
    list_2 = df_2["file_path"].to_list()

    df_3 = pd.read_csv("/Users/ahmed/Desktop/msc-24/TND/kc_classification_data/pre_cutoff_data/df_3.csv")
    list_3 = df_3["file_path"].to_list()

    df_4 = pd.read_csv("/Users/ahmed/Desktop/msc-24/TND/kc_classification_data/pre_cutoff_data/df_4.csv")
    list_4 = df_4["file_path"].to_list()

    return list_1, list_2, list_3, list_4

list_1, list_2, list_3, list_4 = read_data()
```

```python
def intersection_of_lists(list_of_lists):
    if not list_of_lists:
        return []
    # Convert the first list to a set and iteratively intersect with the rest
    result = set(list_of_lists[0])
    for lst in list_of_lists[1:]:
        result.intersection_update(lst)
    return list(result)
```

```python
for i in list_4:
    case_data  = load_json(os.path.join(data_path, i))
    itemid = case_data['itemid']
    articles = case_data['__articles'].split(';')
    similar_cases = []
    for article in articles:
        similar_cases.append(article_map[article])
    similar_cases = intersection_of_lists(similar_cases)
    break
```
