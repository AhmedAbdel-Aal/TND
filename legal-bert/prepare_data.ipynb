```python
import pandas as pd
import os
```

```python
df = pd.read_csv('/home/abdoah/data/test-data/cases_data.csv')
```

```python
test_file_names = list(df['file_path'].values)
```

```python
all_files = os.listdir('/home/abdoah/data/echr-processed')
all_files = list(set(all_files) - set(test_file_names))  
```

```python
# save the list of files to be used for training
with open('/home/abdoah/TND/legal-bert/eval-files.txt', 'w') as f:
    for item in test_file_names:
        f.write("%s\n" % item)
```

```python
# save the list of files to be used for training
with open('/home/abdoah/TND/legal-bert/train-files.txt', 'w') as f:
    for item in all_files:
        f.write("%s\n" % item)
```

```python
# load txt file into a list
def load_files(path):
    with open(path) as f:
        train_files = f.readlines()
        train_files = [x.strip() for x in train_files]
    return train_files
```
