```python
data_path = '/Users/ahmed/Desktop/msc-24/ECHR/echr-processed'
```

```python
import os
files = os.listdir(data_path)

def load_json(file):
    import json
    with open(file) as f:
        return json.load(f)
    

```

```python
years = {}
for idx, file in enumerate(files):
    path = os.path.join(data_path, file)
    data = load_json(path)
    if int(data['importance']) == 2:
        date = data['judgementdate']
        year = date.split('/')[-1]
        year = int(year)
        if year in years:
            years[year] += 1
        else:
            years[year] = 1
```

```python
# cumulative sum
cumulative = {}
for year in sorted(years.keys()):
    cumulative[year] = sum([years[y] for y in years if y <= year])
```
